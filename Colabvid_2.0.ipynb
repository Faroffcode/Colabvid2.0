## ▶️ Start Colabvid in background mode

```python
!mkdir -p logs
!if [ -f .colabvid.pid ] && kill -0 $(cat .colabvid.pid) 2>/dev/null; then echo 'Colabvid is already running.'; else nohup python -u app.py > logs/colabvid.log 2>&1 & echo $! > .colabvid.pid; echo "Colabvid started with PID $(cat .colabvid.pid)"; fi
```
